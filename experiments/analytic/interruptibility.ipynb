# Discussion of Trolley problem
(see `math/2_deterministic.typ`)

```python
from dataclasses import dataclass
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from rich import print


@dataclass
class Interruptibility:
    mi: int = 0
    mt: int = 1
    p: float = 0.5


def Q_keep(e: Interruptibility):
    return -(e.p / (e.mi + 1) + (1 - e.p) / (e.mi + e.mt + 1))


def Q_cut(e: Interruptibility):
    return -(1 / (e.mt + 1))


# Examples

# only task
only_task = Interruptibility(mi=0, mt=1, p=0.5)
print(f"Only task env: Q_keep:{Q_keep(only_task)}, Q_cut: {Q_cut(only_task)}")

both = Interruptibility(mi=1, mt=1, p=0.5)
print(f"When both goals are present: Q_keep:{Q_keep(both)}, Q_cut: {Q_cut(both)}")

high_prob = Interruptibility(mi=1, mt=1, p=0.999)
print(
    f"mi = mt, keep always preferred: Q_keep:{Q_keep(high_prob)}, Q_cut: {Q_cut(high_prob)}"
)
```
